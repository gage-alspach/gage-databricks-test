```python
import csv
import random
import time
import uuid
from datetime import datetime, timezone

output_path = "/Volumes/workspace/gagealspach/ams_demo_source/gateway_source"

locations = ["Reno", "Sparks", "Sacramento", "Fernley", "Carson City"]
customers = ["Acme", "WidgetCo", "Globex", "Initech", "Umbrella"]

while True:
    now = datetime.now(timezone.utc)
    filename = f"generated_{now:%Y%m%d_%H%M%S}_{uuid.uuid4().hex[:6]}.csv"
    filepath = f"{output_path}/{filename}"

    rows = []

    base_version = int(time.time() * 1000)

    for i in range(25):
        rows.append({
            "shipment_id": f"S{uuid.uuid4().hex[:10].upper()}",
            "status": "IN_TRANSIT",
            "location": random.choice(locations),
            "customer_name": random.choice(customers),
            "_source_version": base_version + i,
            "_operation": "INSERT",
            "_source_updated_ts": now.strftime("%Y-%m-%dT%H:%M:%S"),
        })

    with open(filepath, "w", newline="") as f:
        writer = csv.DictWriter(f, fieldnames=rows[0].keys())
        writer.writeheader()
        writer.writerows(rows)

    print(f"Wrote {filename}: {len(rows)} rows")

    time.sleep(10)
```
