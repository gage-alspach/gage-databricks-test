```python
from pyspark import pipelines as dp
from pyspark.sql import functions as F

@dp.materialized_view(name="shipments_gold")
def shipments_gold():
    realtime = spark.read.table("shipments_stream_silver")
    batch = spark.read.table("shipments_batch_silver")

    return (
        realtime
        .join(batch, "shipment_id", "left")
        .select(
            F.col("shipment_id").alias("ShipmentId"),
            F.col("status").alias("Status"),
            F.col("location").alias("CurrentLocation"),
            F.col("customer_name").alias("CustomerName"),
            F.col("service_level").alias("ServiceLevel"),
            F.col("equipment_type").alias("EquipmentType"),
            F.col("origin_terminal").alias("OriginTerminal"),
            F.col("destination_terminal").alias("DestinationTerminal"),
            F.col("weight_lbs").alias("WeightLbs"),
        )
    )
```
